```python
df = spark.read.format("csv").option("header","true").load("/Workspace/Users/karthikkolluru123@gmail.com/df.csv")
df.show()

# selectExpr does select as well as expressions
print('===upper case category===')
df.selectExpr("product","tdate"," upper(category) as category ").show()

# increment amount and uppercase category
selExpr = df.selectExpr("id",
                        "tdate",
                        "cast(amount as double) + 150 as amountinc",
                        " upper(category) as category ",
                        "product",
                        "spendby")
print('===uppercase category case category, amount increment===')
selExpr.show()

# split tdate and get only year
print('===split tdate and get only year===')
dataExpr = df.selectExpr(
    "id",
    "split(tdate, '-')[2] as year",
    "cast(amount as double) + 150 as amountinc",
    " upper(category) as category ",
    "concat(product,'~zeyo') as product",
    "spendby"
)
dataExpr.show()
```
