```python
df = spark.read.format("csv").option("header","true").load("/Workspace/Users/karthikkolluru123@gmail.com/df.csv")

fildf = df.filter(df.category == 'Exercise')
fildf2 = df.filter(" category = 'Exercise' ")
df.show()
print('===filter by exercise===')
fildf.show()
print('===filter by exercise using df.category===')
fildf2.show()

andfil = df.filter(" category = 'Exercise' and spendby = 'cash' ")
print('===filter by exercise and mode===')
andfil.show()

orfil = df.filter(" category = 'Exercise' or spendby = 'cash' ")
print('===filter by exercise or mode===')
orfil.show()

infil = df.filter(" category in ('Exercise','Gymnastics')")
print('===filter using in Exercise Gymnastics===')
infil.show()

notinfil = df.filter(" category not in ('Exercise','Gymnastics')")
print('===filter using not in===')
notinfil.show()

likefil =  df.filter(" product like '%Gymnastics%'")
print('===filter using like Gymnastics===')
likefil.show()

notnullfil = df.filter(" product is not null")
print('===filter using product not null===')
notnullfil.show()

isnullfil = df.filter(" product is null")
print('===filter using product is null===')
isnullfil.show()

dropfil = df.drop("spendby")
print('===drop spendby===')
dropfil.show()
```
