```python
import pandas as pd
from datetime import datetime, timedelta
import random

# Set seed for reproducibility
random.seed(42)

# Generate data
data = []
for i in range(1, 21):
    txnid = f"{i:06d}"  # Format as 6-digit string with leading zeros
    txndate = (datetime(2024, 1, 1) + timedelta(days=random.randint(0, 60))).strftime('%Y-%m-%d')
    amount = round(random.uniform(10, 500), 2)
    # Define category-product mapping
    category_products = {
        'Field': ['Javelin', 'Shot Put', 'Discus', 'Hurdles', 'Running Spikes'],
        'Exercise': ['Treadmill', 'Dumbbells', 'Yoga Mat', 'Resistance Bands', 'Exercise Ball'],
        'Gymnastics': ['Rings', 'Balance Beam', 'Parallel Bars', 'Vault', 'Floor Mat'],
        'Aquatics': ['Goggles', 'Swim Cap', 'Kickboard', 'Pull Buoy', 'Fins'],
        'Combat Sports': ['Boxing Gloves', 'Punching Bag', 'Hand Wraps', 'Mouth Guard', 'Shin Guards']
    }
    
    category = random.choice(list(category_products.keys()))
    product = random.choice(category_products[category])
    mode = random.choice(['Cash', 'Credit', 'Debit', 'UPI', 'Net Banking'])
    
    data.append({
        'txnid': txnid,
        'txndate': txndate,
        'amount': amount,
        'category': category,
        'product': product,
        'mode': mode
    })

# Create DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)
print(f"\nDataFrame shape: {df.shape}")
```
